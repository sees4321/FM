```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 논문용 깔끔한 스타일 설정
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'serif',            # 폰트 스타일을 명조(Serif) 계열로 설정
    'font.serif': ['Times New Roman'], # 1순위 폰트를 Times New Roman으로 지정
    'font.size': 16,
    'axes.labelsize': 20,
    'axes.titlesize': 20,
    'legend.fontsize': 16,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'pdf.fonttype': 42, # 폰트가 PDF에 내장되도록 설정 (논문 제출 필수 요건)
    'ps.fonttype': 42
})


tuab = np.array([0.7834, 0.7860, 0.7757, 0.7693, 0.7487])
tuab_err = np.array([.004, .0024, .0019, .0054, .0033])
tuabB = np.array([0.7825, 0.7839, 0.7686, 0.7609, 0.7549])
tuabB_err = np.array([.005498, .0065, .0034, .0023, .0055])

isruc = np.array([0.7105, 0.7206, 0.7124, 0.6830, 0.6514])
isruc_err = np.array([.022, .024, .019, .017, .027])
isrucB = np.array([0.7093, 0.7123, 0.6324, 0.6409, 0.5998])
isrucB_err = np.array([.020, .018436, .013, .017, .015])

mi = np.array([0.5019, 0.5227, 0.4659, 0.4293, 0.4443])
mi_err = np.array([.017, .011, .022, .013, .017])
miB = np.array([0.4708, 0.4854, 0.4487, 0.4298, 0.4233])
miB_err = np.array([.009, .015, .018, .011, .010])

markers = ['^', 'o']
colors = ['#2B79D3', '#E66466']

plt.figure(figsize=(18, 6))
plt.subplot(133)
plt.plot(tuabB, label='Block', color=colors[0], marker=markers[0])
# plt.errorbar(range(len(tuabB)), tuabB, yerr=tuabB_err, fmt='-o', capsize=5)
plt.fill_between(range(len(tuabB)), [a - b for a, b in zip(tuabB, tuabB_err)], [a + b for a, b in zip(tuabB, tuabB_err)], color=colors[0], alpha=0.2)#, label='Error Range')
plt.plot(tuab, label='SSP', color=colors[1], marker=markers[1])
# plt.errorbar(range(len(tuab)), tuab, yerr=tuab_err, fmt='-o', capsize=5)
plt.fill_between(range(len(tuab)), [a - b for a, b in zip(tuab, tuab_err)], [a + b for a, b in zip(tuab, tuab_err)], color=colors[1], alpha=0.2)# , label='Error Range')
plt.xticks(range(len(mi)), ['0.27', '0.34', '0.45', '0.65', '0.75'])
plt.xlabel('Expected Masking Ratio')
# plt.ylabel('Accuracy')
plt.title('TUAB')
plt.legend()
# plt.grid()

plt.subplot(132)
plt.plot(isrucB, label='ISRUC+B', color=colors[0], marker=markers[0])
# plt.errorbar(range(len(isrucB)), isrucB, yerr=isrucB_err, fmt='-o', capsize=5)
plt.fill_between(range(len(isrucB)), [a - b for a, b in zip(isrucB, isrucB_err)], [a + b for a, b in zip(isrucB, isrucB_err)], color=colors[0], alpha=0.2, label='Error Range')
plt.plot(isruc, label='ISRUC', color=colors[1], marker=markers[1])
# plt.errorbar(range(len(isruc)), isruc, yerr=isruc_err, fmt='-o', capsize=5)
plt.fill_between(range(len(isruc)), [a - b for a, b in zip(isruc, isruc_err)], [a + b for a, b in zip(isruc, isruc_err)], color=colors[1], alpha=0.2, label='Error Range')
plt.xticks(range(len(mi)), ['0.27', '0.34', '0.45', '0.65', '0.75'])
plt.xlabel('Expected Masking Ratio')
# plt.ylabel('Accuracy')
plt.title('ISRUC')
# plt.grid()

plt.subplot(131)
plt.plot(miB, label='MI+B', color=colors[0], marker=markers[0])
# plt.errorbar(range(len(miB)), miB, yerr=miB_err, fmt='-o', capsize=5)
plt.fill_between(range(len(miB)), [a - b for a, b in zip(miB, miB_err)], [a + b for a, b in zip(miB, miB_err)], color=colors[0], alpha=0.2, label='Error Range')
plt.plot(mi, label='MI', color=colors[1], marker=markers[1])
# plt.errorbar(range(len(mi)), mi, yerr=mi_err, fmt='-o', capsize=5)
plt.fill_between(range(len(mi)), [a - b for a, b in zip(mi, mi_err)], [a + b for a, b in zip(mi, mi_err)], color=colors[1], alpha=0.2, label='Error Range')
plt.xticks(range(len(mi)), ['0.27', '0.34', '0.45', '0.65', '0.75'])
plt.xlabel('Expected Masking Ratio')
plt.ylabel('Balanced Accuracy')
plt.title('PhysionNetMI')
# plt.grid()

plt.tight_layout()
plt.savefig('fig_masking.pdf', format='pdf', dpi=300, bbox_inches='tight')
# sns.despine()
plt.show()
```
