```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'font.size': 14,
    'axes.labelsize': 16,
    'axes.titlesize': 16,
    'legend.fontsize': 13,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'pdf.fonttype': 42,
    'ps.fonttype': 42
})

# Model size scaling: 20M, 45M, 200M, 400M
model_sizes = np.array([20, 45, 200, 400])
tuab = np.array([.584189, .601467, .66284, .683789])
tuab_err = np.array([.006, .005, .004662, .005])

# Data scaling: 10%, 25%, 50%, 100% (using e.g., 200M model)
data_fracs = np.array([10, 25, 50, 100])
# isruc = np.array([.516307, .565122, .569637, .601467])
# isruc_err = np.array([.007403, .010403, .01506, .005])
# isruc = np.array([.586307, .625122, .639637, .66284])
# isruc_err = np.array([.007403, .010403, .01506, .004662])
isruc = np.array([.51, .56, .61, .66284])
isruc_err = np.array([.004662, .004662, .004662, .004662])

# Baseline accuracies (예시 - 실제 값으로 교체)
baseline_a = 0.496
baseline_b = 0.610 
baseline_c = 0.648 

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Left: Model Size Scaling ---
ax = axes[0]

# ax.axhline(baseline_a, linestyle='--', color='red', linewidth=1.2)
# ax.text(model_sizes[1], baseline_a + 0.003, 'CbraMod (4M)', fontsize=11, ha='right', color='red')
ax.axhline(baseline_b, linestyle='--', color='#145A32', linewidth=1.2)
ax.text(model_sizes[1], baseline_b + 0.003, 'REVE-Base (67M)', fontsize=11, ha='right', color='#145A32')
ax.axhline(baseline_c, linestyle='--', color='#D55E00', linewidth=1.2)
ax.text(model_sizes[1], baseline_c + 0.003, 'REVE-Large (408M)', fontsize=11, ha='right', color='#D55E00')

ax.plot(model_sizes, tuab, marker='o', linewidth=2, markersize=8,
        label='Ours', color="#1A365D")
ax.fill_between(model_sizes, tuab - tuab_err, tuab + tuab_err,
                alpha=0.2, color="#1A365D")
# 2E86AB
ax.set_xscale('log')
ax.set_xticks(model_sizes)
ax.set_xticklabels(['20M', '45M', '200M', '400M'])
ax.minorticks_off()
ax.set_xlabel('Model Size')
ax.set_ylabel('Average Balanced Accuracy')
ax.set_title('Model Size Scaling')
ax.legend(loc='lower right')

# --- Right: Data Scaling ---
ax = axes[1]

# ax.axhline(baseline_a, linestyle='--', color='red', linewidth=1.2)
# ax.text(data_fracs[1], baseline_a + 0.003, 'CbraMod (4M)', fontsize=11, ha='right', color='red')
ax.axhline(baseline_b, linestyle='--', color='#145A32', linewidth=1.2)
ax.text(data_fracs[2], baseline_b + 0.003, 'REVE-Base (67M)', fontsize=11, ha='right', color='#145A32')
ax.axhline(baseline_c, linestyle='--', color='#D55E00', linewidth=1.2)
ax.text(data_fracs[1], baseline_c + 0.003, 'REVE-Large (408M)', fontsize=11, ha='right', color='#D55E00')

ax.plot(data_fracs, isruc, marker='o', linewidth=2, markersize=8,
        label='Ours (200M)', color='#1A365D')
ax.fill_between(data_fracs, isruc - isruc_err, isruc + isruc_err,
                alpha=0.2, color='#1A365D')

ax.set_xscale('log')
ax.set_xticks(data_fracs)
ax.set_xticklabels(['10%', '25%', '50%', '100%'])
ax.minorticks_off()
ax.set_xlabel('Pre-training Data Fraction')
ax.set_ylabel('Average Balanced Accuracy')
ax.set_title('Data Scaling')
ax.legend(loc='lower right')

plt.tight_layout()
plt.savefig('fig_scaling.pdf', format='pdf', dpi=300, bbox_inches='tight')
plt.show()
```
